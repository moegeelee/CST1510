```text
# --- TRY IT ------------------------------------------------------------------

```

````python
percent = 105
if percent >= 100:
    print("OVER LIMIT")
elif percent >= 90:
    print("WARNING")
else:
    print("OK")
attempts = 3
max_attempts = 3
if attempts >= max_attempts:
    print("LOCKED")
else:
    print("OK")
# Extended version: the WARNING case is exactly one below the maximum.
if attempts >= max_attempts:
    print("LOCKED")
elif attempts == max_attempts - 1:
    print("WARNING")
else:
    print("OK")
````

## 02_sentinel_and_running_total.py

Source: `03 - while Loops/examples/02_sentinel_and_running_total.py`

```text
# --- TRY IT ------------------------------------------------------------------
# 1. Adapt section 2 so it also counts how many numbers were entered.
# 2. Adapt section 3 so it stops on "quit" instead of after exactly 3 checks.
# 3. In section 4, change the condition so it skips every EVEN number instead.

```

````python
total = 0.0
number_count = 0
while True:
    entry = input("Number or done: ")
    if entry == "done":
        break
    total += float(entry)
    number_count += 1
print("Count:", number_count)
print("Total:", total)
````

````python
over_limit_count = 0
while True:
    entry = input("Value or quit: ")
    if entry == "quit":
        break
    value = float(entry)
    limit = float(input("Limit: "))
    if value > limit:
        over_limit_count += 1
print("Over limit:", over_limit_count)
````

````python
number = 0
while number < 5:
    number += 1
    if number % 2 == 0:
        continue
    print(number)
````

```text
# 3. Print the numbers 20 down to 0, counting by 4s.

```

````python
for number in range(1, 6):
    print(f"Record {number}")
for number in range(5, 51, 5):
    print(number)
for number in range(20, -1, -4):
    print(number)
````

## 02_for_vs_while.py

Source: `04 - for Loops and range/examples/02_for_vs_while.py`

```text
# --- TRY IT ------------------------------------------------------------------
# 1. Which tool would you use to print the 12 times table? Write it.
# 2. Which tool would you use to keep retrying a password until it is
#    correct? Write it.
# 3. Rewrite section 1 as a while loop that also works for any number of
#    records, not just exactly 3.

```

````python
# Twelve repetitions are known in advance: for.
for number in range(1, 13):
    print(f"12 x {number} = {12 * number}")
# The number of password attempts is unknown: while.
while True:
    password = input("Password: ")
    if password == "python123":
        break
````

````python
# Ask for the count so the while loop can handle any chosen batch size.
record_count = int(input("How many records? "))
record_number = 0
while record_number < record_count:
    label = input("Record ID: ")
    value = float(input("Value: "))
    limit = float(input("Limit: "))
    if value > limit:
        print(f"{label}: OVER LIMIT")
    else:
        print(f"{label}: OK")
    record_number += 1
````
