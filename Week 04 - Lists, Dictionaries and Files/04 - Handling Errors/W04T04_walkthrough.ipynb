## 2. `try` / `except` — catch it, decide what happens next

Python runs the `try` block. If the named error happens inside it, execution
jumps straight to the matching `except` — the rest of the `try` block is
skipped, but the program does **not** crash.

````python
try:
    with open("missing.txt") as f:
        text = f.read()
except FileNotFoundError:
    print("missing.txt not found — using an empty report instead")
    text = ""

print(repr(text))     # '' — the program kept going
````

## 3. Name the exception you expect

`except:` on its own catches **everything** — including a typo in your own
code, three lines away, that has nothing to do with the problem you were
actually planning for. Name the exception. If more than one thing can go
wrong, use more than one `except`.

````python
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        print("can't divide by zero")
        return None
    except TypeError:
        print("both values need to be numbers")
        return None

print(safe_divide(10, 2))      # 5.0
print(safe_divide(10, 0))      # can't divide by zero -> None
print(safe_divide(10, "two"))  # both values need to be numbers -> None
````

```python
# WATCH OUT — a bare except hides real bugs, it does not fix them
try:
    result = 10 / 0
except:                    # catches EVERYTHING — even a typo you made
    print("something went wrong")   # ...and you will never know what
```

`as e` gives you the actual error object, if you want to see the real message
rather than writing your own:

```python
try:
    int("abc")
except ValueError as e:
    print("caught it:", e)     # caught it: invalid literal for int() with base 10: 'abc'
```

### Your turn
Write a function that takes a filename and a fallback value. Inside, `try` to open and read the file; `except FileNotFoundError`, print a message and return the fallback instead.

````python
def read_or_default(filename, fallback):
    """Return the file contents, or the fallback if the file is missing."""
    try:
        with open(filename) as f:
            return f.read()
    except FileNotFoundError:
        print(f"{filename} not found; using the fallback")
        return fallback

with open("available.txt", "w") as f:
    f.write("Record loaded")
print(read_or_default("available.txt", "No records"))
print(read_or_default("not_here.txt", "No records"))
````

## Topic 4 Challenge

Write `safe_load(path)`: it should `try` to open and read `path`, returning the
contents. If the file does not exist, it should print a clear one-line message
and return `""` instead — never let it crash the program that calls it.

````python
def safe_load(path):
    """Return file contents, or an empty string if the file is missing."""
    try:
        with open(path) as f:
            return f.read()
    except FileNotFoundError:
        print(f"{path} not found; returning empty text")
        return ""

print(safe_load("available.txt"))
print(repr(safe_load("not_here.txt")))
````

## Same idea, three fields

A program that crashes on the first bad input is not "correct" — it is just early to quit.

| Field | Looks like |
|---|---|
| **AI / Data Science** | `except FileNotFoundError:` — skip a dataset that hasn't been uploaded yet, keep processing the rest |
| **Cyber Security** | `except ValueError:` — skip a malformed log line, keep scanning the rest of the file |
| **IT** | `except FileNotFoundError:` — fall back to default settings when a config file is missing |

Same shape, three domains: catch it, decide what "carry on" means, keep going.

---
## Common mistakes

| Mistake | What you see | Fix |
|---|---|---|
| a bare `except:` | it catches EVERYTHING, including your own typos | always name the exception: `except ValueError:` |
| catching the wrong exception | the error still crashes the program | read the traceback's **last line** — that word IS the exception name to catch |
| catching an error and doing nothing | the program looks like it worked, but the data is just quietly missing | at minimum, print what happened — a silent `except: pass` is the hardest bug to find |
| putting too much inside one `try` | you can no longer tell which line actually failed | keep the `try` block to just the risky line or two |

## What the handler covers

`except FileNotFoundError` handles that named failure only. A permission error or a decoding problem still propagates, so “safe” does not mean every possible error is hidden. The return in an `except` block supplies a usable fallback to the caller; printing a message alone would leave an implicit `None` return.
