## 04 - Handling Errors / 01_catching_errors

1. Write a function that asks for a number with input() and int(), catching
   ValueError and asking again if it fails. (Hint: wrap it in a while loop.)
2. Trigger a FileNotFoundError on purpose, on a file you know doesn't exist.
   Read the traceback - which line does Python say the error happened on?
3. Add a second except block to section 2 that also catches PermissionError.

```python
def ask_integer():
    """Ask for an integer again whenever numeric conversion fails."""
    while True:
        try:
            return int(input("Whole number: "))
        except ValueError:
            print("That is not a whole number; try again.")

print(ask_integer())
```

```python
with open("intentionally_missing.txt") as f:  # this open call raises the error
    print(f.read())
```

```python
def read_with_messages(path):
    """Return file contents or a fallback for missing and inaccessible files."""
    try:
        with open(path) as f:
            return f.read()
    except FileNotFoundError:
        print(f"{path} not found")
        return ""
    except PermissionError:
        print(f"No permission to read {path}")
        return ""

print(repr(read_with_messages("intentionally_missing.txt")))
```

The deliberate failure occurs on the `open(...)` line. In a live notebook its traceback shows that line; the automated saved error is shortened to the exception name and message. Permission handling is checked separately with a simulated PermissionError, without changing permissions on personal files.

## 04 - Handling Errors / 02_a_safer_loader

1. Call safe_load_records on a file you know exists - confirm it still works
   normally, not just when something is missing.
2. Add a third bad line to broken_records.txt (e.g. a value that isn't a
   number) and confirm it gets skipped too, not just lines with the wrong
   number of fields.
3. In your own words: why does section 2 wrap the CALL to load_records,
   instead of putting a try/except inside load_records itself?

```python
def load_records(path):
    """Read label,value,limit lines into a list of dictionaries."""
    records = []
    with open(path) as f:
        for line in f:
            label, value, limit = line.strip().split(",")
            records.append({"label": label, "value": float(value), "limit": float(limit)})
    return records

def safe_load_records(path):
    """Return records or an empty list when the file is missing."""
    try:
        return load_records(path)
    except FileNotFoundError:
        print(f"{path} not found; using an empty list")
        return []

print(safe_load_records("../03 - Files/examples/sample_records.txt"))

with open("broken_records.txt", "w") as f:
    f.write("R-001,87,100\n")
    f.write("wrong shape\n")
    f.write("R-003,45,100\n")
    f.write("R-004,not_a_number,100\n")

def safe_load_records_v2(path):
    """Load valid records, reporting missing files and skipped malformed lines."""
    records = []
    try:
        with open(path) as f:
            for line in f:
                try:
                    label, value, limit = line.strip().split(",")
                    records.append({"label": label, "value": float(value), "limit": float(limit)})
                except ValueError:
                    print("Skipping a bad line:", line.strip())
    except FileNotFoundError:
        print(f"{path} not found; using an empty list")
    return records

print(safe_load_records_v2("broken_records.txt"))
```

Study explanation: wrapping the call keeps the basic loader focused on reading data and lets its caller choose the response to a missing file. One caller can use an empty list; another may ask for a different filename. The more forgiving loader deliberately adopts a different policy: report and skip malformed lines.

## 05 - Putting It Together / 02_load_and_process

1. Add a WARNING tier (90% or more) to the loop in section 2.
2. Write your own load_records-style function for a file with a different
   number of fields per line.
3. Change the loop in section 3 to also count how many were exactly OK.

```python
records = load_records("../05 - Putting It Together/examples/sample_records.txt")
over_limit_count = 0
ok_count = 0
for record in records:
    percent = record["value"] / record["limit"] * 100
    if percent >= 100:
        status = "OVER LIMIT"
        over_limit_count += 1
    elif percent >= 90:
        status = "WARNING"
    else:
        status = "OK"
        ok_count += 1
    print(f"{record['label']}: {status}")
print(f"{over_limit_count} of {len(records)} were over limit")
print(f"{ok_count} of {len(records)} were exactly OK")

with open("properties.txt", "w") as f:
    f.write("Lekki_flat,1400\nIkeja_flat,1000\n")

def load_properties(path):
    """Read name,rent lines into dictionaries with two fields."""
    properties = []
    with open(path) as f:
        for line in f:
            name, rent = line.strip().split(",")
            properties.append({"name": name, "rent": float(rent)})
    return properties

print(load_properties("properties.txt"))
```
