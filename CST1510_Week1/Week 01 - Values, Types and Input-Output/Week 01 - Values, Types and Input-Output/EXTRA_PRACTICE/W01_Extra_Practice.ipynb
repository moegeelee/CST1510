## 02_calculations.py

Source: `03 - Numbers and Arithmetic/examples/02_calculations.py`

```text
# --- TRY IT ----------------------------------------------------------------

#    percentage that failed, and the percentage that passed.
# 2. Do it without typing any number you could calculate.
# 3. Start total at 100. Subtract 30, then double it, using short forms.

```

````python
records_total = 250
records_failed = 38
records_passed = records_total - records_failed
print(records_passed)
print((records_failed / records_total) * 100)
print((records_passed / records_total) * 100)
total = 100
total -= 30
total *= 2
print(total)
````

## 02_fstrings.py

Source: `04 - Input, Casting and f-strings/examples/02_fstrings.py`

```text
# --- TRY IT ----------------------------------------------------------------
# 1. Print 7 / 3 to exactly three decimal places.
# 2. Print your name left-aligned in 15 characters, then a number right-
#    aligned in 8. Do it for three different names and check they line up.
# 3. Remove the f from one line above. Run it. Put it back.

```

````python
print(f"{7 / 3:.3f}")
# The course width of 15 is a minimum; a longer full name is not cut off.
student_name = "MOHAMMED"
print(f"{student_name:<15}{87:>8}")
print(f"{'Amina':<15}{120:>8}")
print(f"{'David':<15}{45:>8}")
print("{student_name:<15}{87:>8}")  # Missing f: literal braces.
print(f"{student_name:<15}{87:>8}")
````
