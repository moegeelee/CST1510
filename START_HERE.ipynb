## Lesson 1 — a computer follows exact instructions

Think of a tiny Nairaprop rent summary. This example uses invented numbers in arbitrary currency units and does not access your real platform.

A **value** is a piece of data, such as `2500` or `"Lagos"`. A **variable** is a name attached to a value. A **statement** is an instruction. Python runs these statements from top to bottom.

Read `monthly_rent = 2500` as “monthly_rent gets 2500.” It stores a value; it is not asking a question. `print()` shows a result on screen.

Before running the next cell, predict the two output lines.

```python
monthly_rent = 2500
months = 12
annual_rent = monthly_rent * months
print(monthly_rent)
print(annual_rent)
```

The first output is `2500`. The second is `30000`.

Trace the instructions: store 2500 → store 12 → multiply those values → store 30000 → display the two requested values.

`annual_rent` holds the calculated result at that moment. If you later change `monthly_rent`, Python does not automatically recalculate `annual_rent`. You would run the multiplication again.

**Try it yourself:** change the monthly rent to 1800 and the number of months to 6. Predict the total before running. Explain why using the variable names is better than typing the total into the program.

## Lesson 2 — text and numbers behave differently

Quotation marks matter. `"20"` is text (a string); `20` is a whole number (an integer). The `+` operator joins strings but adds numbers.

First predict both outputs. Then run the cell and explain why they differ.

```python
print("20" + "5")
print(20 + 5)
print(type("20"))
print(type(20))
```

`input()` returns text. If a person types `2500`, the program receives `"2500"`. Convert it before doing arithmetic. Read the next expression inside out:

1. `input(...)` asks and receives text.
2. `float(...)` converts numeric text to a decimal number.
3. `monthly_rent = ...` stores that number.

Enter a number, such as 2500.

```python
monthly_rent = float(input("Monthly rent: "))
annual_rent = monthly_rent * 12
print(f"Annual rent: {annual_rent:.2f}")
```

In the last line, `f` enables values inside braces. `.2f` displays exactly two digits after the decimal point. Formatting changes the display; it does not make a calculation more accurate.

## Precision notes from the review

- `print` is a built-in name, not a reserved keyword. Overwriting it is legal but causes confusion. [Python built-ins](https://docs.python.org/3/library/functions.html).
- `//` rounds down; for negative numbers it is not the same as discarding a fraction. [Python arithmetic](https://docs.python.org/3/reference/expressions.html#binary-arithmetic-operations).
- A local parameter is a name bound to an object, not an automatic independent copy. Functions can read globals and mutate shared objects; local scope alone does not make a function harmless. We use explicit parameters and returns in this course. [Python names and scope](https://docs.python.org/3/tutorial/classes.html#python-scopes-and-namespaces).
