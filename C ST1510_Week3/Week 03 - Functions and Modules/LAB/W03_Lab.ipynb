### D5. Arguments are matched by position

2. Print `within_limit(87, 100)`, then `within_limit(100, 87)`.

### D6. A default value

1. Define `status_of(percent, warning_at=90)` that returns:
   - `"OVER LIMIT"` if `percent` is 100 or more
   - `"WARNING"` if `percent` is `warning_at` or more
   - `"OK"` otherwise
2. Print `status_of(95)`, then `status_of(85)`, then `status_of(85, 80)`.

**Expected output:** `WARNING`, `OK`, `WARNING`

````python
def status_of(percent, warning_at=90):
    if percent >= 100:
        return "OVER LIMIT"
    elif percent >= warning_at:
        return "WARNING"
    else:
        return "OK"


print(status_of(95))
print(status_of(85))
print(status_of(85, 80))
````

### D8. `*args` and `**kwargs`

a) Define `total(*numbers)` that returns the sum of all the numbers passed in. Print `total(1, 2, 3)`,
   `total(10, 20, 30, 40)` and `total()`.

b) Define `configure(**settings)` that loops over `settings.items()` and prints each one as `key = value`.
   Call it with `configure(host="10.0.0.1", port=8080)`.

**Expected output:**
```
6

0

```

````python
def total(*numbers):
    return sum(numbers)


print(total(1, 2, 3))
print(total(10, 20, 30, 40))
print(total())


def configure(**settings):
    for key, value in settings.items():
        print(f"{key} = {value}")


configure(host="10.0.0.1", port=8080)
````

### C1. Wire two functions together

Using `percent_of` from D3 and `status_of` from D6 **together**, print a status for each of these
three pairs. Do not write the percentage or status logic a second time.

| value | limit | Expected output |
|---|---|---|
| 87 | 100 | `OK` |
| 120 | 100 | `OVER LIMIT` |
| 45 | 100 | `OK` |

````python
def percent_of(part, total):
    """Return part as a percentage of total."""
    return (part / total) * 100


def status_of(percent, warning_at=90):
    if percent >= 100:
        return "OVER LIMIT"
    elif percent >= warning_at:
        return "WARNING"
    else:
        return "OK"


for value, limit in [(87, 100), (120, 100), (45, 100)]:
    percentage = percent_of(value, limit)
    print(status_of(percentage))
````

### C2. How many are over?

Write `count_over(*values, limit=100)` with a one-line docstring. It returns how many of the values
are over the limit.

| Call | Expected output |
|---|---|
| `count_over(87, 120, 45, 101)` | `2` |
| `count_over(87, 120, 45, 101, limit=50)` | `3` |
| `count_over()` | `0` |

````python
def count_over(*values, limit=100):
    """Return how many values are over the limit."""
    count = 0
    for value in values:
        if value > limit:
            count += 1
    return count


print(count_over(87, 120, 45, 101))
print(count_over(87, 120, 45, 101, limit=50))
print(count_over())
````

## My own learning notes — what I understood and how I solved it

I learned that a function is just a reusable block of code. Instead of writing the same logic over and over, I can define it once, give it a name, and call it whenever I need it. For example, `status_of(percent)` handles the logic for deciding if a result is OK, WARNING, or OVER LIMIT. This makes the program cleaner and easier to test.

The biggest idea this week was the difference between `print()` and `return`.

- `print()` shows a value on the screen.
- `return` sends a value back to where the function was called.

This matters because if I only print inside a function, I cannot use that value elsewhere. If I return a value, I can store it in a variable and use it again. That is why I used `return` in `percent_of`, `status_of`, and `check`.

I also learned that variables inside a function are local. A variable made inside a function does not exist outside it. That is why the function `make_status()` could print `result` inside itself, but `print(result)` outside the function caused a `NameError`. To move data out of a function, I must `return` it.

Arguments and parameters were another key idea. A parameter is the name inside the function, while an argument is the value passed in when I call the function. I used position matching in `within_limit(87, 100)`, and I saw that changing the order changes the result. I also used default parameters like `warning_at=90`, so a function can still work if I do not pass that value.

The keyword argument section helped me understand that `name=value` makes it clearer and avoids confusion. `status_of(percent=95, warning_at=96)` works because the names match the parameter names. I also learned that Python does not allow a positional value after a keyword argument, so `status_of(percent=95, 96)` is invalid.

For `*args` and `**kwargs`, I understood that:

- `*args` collects extra positional values into a tuple.
- `**kwargs` collects extra named values into a dictionary.

That is why `total(*numbers)` can accept any number of values and add them all together. `configure(**settings)` can accept named settings and print each key and value.

Modules also became clearer. `import math` gives me access to functions like `sqrt()` and `pi`. `from math import sqrt` lets me use it directly without writing `math.` every time. I also learned that the `__name__` variable is different when a file is run directly compared with when it is imported. The code inside `if __name__ == "__main__":` only runs when I run the file directly, which is useful for test code or demonstration code.
