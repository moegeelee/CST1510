## 02_why_functions.py

Source: `01 - Defining and Calling Functions/examples/02_why_functions.py`

```text
# --- TRY IT ------------------------------------------------------------------
# 1. Name one reason percent_of and status_of are two functions and not one.
# 2. Write a function is_valid_percent(p) that returns True if p is between
#    0 and 100 inclusive. Use it to check percent_of(87, 100).

```

````python
# Separate calculation from policy: a threshold change need not change the formula.
def percent_of(part, total):
    """Return a percentage for a nonzero total."""
    return (part / total) * 100

def is_valid_percent(p):
    """Return whether p is in the closed interval from 0 to 100."""
    return (p >= 0) and (p <= 100)

print(is_valid_percent(percent_of(87, 100)))
````

## 01_multiple_and_default.py

Source: `02 - Parameters and Arguments/examples/01_multiple_and_default.py`

```text
# --- TRY IT ------------------------------------------------------------------
# 1. Write report_line(label, value, width=10) that returns label followed by
#    value right-aligned in width characters. Call it once using the default
#    width and once overriding it.
# 2. Predict, then check: status_of(89) and status_of(89, 85)

```

````python
def report_line(label, value, width=10):
    """Return a label and a number aligned to the requested width."""
    return f"{label}: {value:>{width}}"

def status_of(percent, warning_at=90):
    """Return the status using the supplied warning threshold."""
    if percent >= 100:
        return "OVER LIMIT"
    elif percent >= warning_at:
        return "WARNING"
    return "OK"

print(report_line("Used", 87))
print(report_line("Used", 87, width=15))
# Worked predictions: OK, then WARNING.
print(status_of(89))
print(status_of(89, 85))
````

## 03_args_and_kwargs.py

Source: `02 - Parameters and Arguments/examples/03_args_and_kwargs.py`

```text
# --- TRY IT ------------------------------------------------------------------
# 1. Write average(*values) that returns the mean of however many numbers it
#    gets. Call it with 3 numbers, then with 6.
# 2. Write log_event(event, **details) that prints the event name, then one
#    line per extra detail passed in as a keyword argument.
# 3. Call connect() from section 4 with a host and three keyword options of
#    your own choosing - predict the output before running it.

```

````python
def average(*values):
    """Return the mean, or None when there are no values."""
    if len(values) == 0:
        return None
    return sum(values) / len(values)

def log_event(event, **details):
    """Print an event and its named details."""
    print(event)
    for key, value in details.items():
        print(f"{key} = {value}")

def connect(host, *extra_hosts, **options):
    """Print the hosts and connection options for this demonstration."""
    print("primary host:", host)
    if extra_hosts:
        print("also connecting to:", extra_hosts)
    for key, value in options.items():
        print(f"  option {key} = {value}")

print(average(3, 6, 9))
print(average(1, 2, 3, 4, 5, 6))
log_event("dataset_checked", dataset="survey_2026", rows=1187)
# Worked prediction: host, then timeout=5, retries=3, ssl=True in that order.
connect("10.0.0.1", timeout=5, retries=3, ssl=True)
````
